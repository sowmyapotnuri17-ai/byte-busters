```python

import gradio as gr
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity
from sklearn.pipeline import Pipeline
from sklearn.linear_model import LogisticRegression

# =====================================================
# 1. GOVERNMENT SCHEME DATA
# =====================================================

SCHEMES = [
    {
        "name": "National Scholarship Portal (NSP)",
        "aliases": "scholarship student education NSP",
        "category": "Education",
        "description": "The National Scholarship Portal provides access to various Central Government and State Government scholarships.",
        "eligibility": "Eligibility depends on the individual scholarship. Conditions can include education level, marks, family income, domicile and scheme-specific rules.",
        "age": "There is no single age limit for all scholarships. Check the selected scholarship's official guidelines.",
        "documents": [
            "Academic certificates and marksheets",
            "Income certificate, if required",
            "Domicile certificate, if required",
            "Identity details required by the scheme",
            "Bank account details, if required"
        ],
        "benefits": "Financial assistance according to the selected scholarship's rules.",
        "application": "Visit the National Scholarship Portal, select a scholarship and follow its official application instructions.",
        "url": "https://scholarships.gov.in/"
    },
    {
        "name": "PM-KISAN Samman Nidhi",
        "aliases": "PM KISAN farmer agriculture farming land payment",
        "category": "Agriculture",
        "description": "An income-support scheme for eligible landholding farmer families in India.",
        "eligibility": "Eligibility depends on landholding records, scheme conditions and exclusions. Not every farmer automatically qualifies.",
        "age": "Check the current official guidelines. Age alone does not determine eligibility.",
        "documents": [
            "Aadhaar details",
            "Landholding records",
            "Bank account details",
            "Other records requested by authorities"
        ],
        "benefits": "Eligible farmer families receive Rs. 6,000 per year in three equal instalments, subject to scheme rules.",
        "application": "Visit the official PM-KISAN portal to review registration instructions and beneficiary status.",
        "url": "https://pmkisan.gov.in/"
    },
    {
        "name": "Ayushman Bharat PM-JAY Senior Citizens",
        "aliases": "Ayushman Bharat PMJAY healthcare hospital medical senior citizen elderly 70",
        "category": "Healthcare",
        "description": "The senior-citizen expansion of Ayushman Bharat PM-JAY provides health coverage to eligible people aged 70 years and above under applicable provisions.",
        "eligibility": "The senior-citizen provision covers people aged 70 years and above, subject to official enrolment procedures.",
        "age": "70 years and above for the senior-citizen provision.",
        "documents": [
            "Proof of age and identity",
            "Aadhaar details as required for enrolment",
            "Other information requested by the official system"
        ],
        "benefits": "Health coverage up to Rs. 5 lakh per year on a family basis under applicable senior-citizen provisions.",
        "application": "Check the official National Health Authority website for current eligibility and enrolment instructions.",
        "url": "https://nha.gov.in/"
    },
    {
        "name": "Pradhan Mantri Mudra Yojana (PMMY)",
        "aliases": "MUDRA loan business small business self employment entrepreneur shop",
        "category": "Business and Self-employment",
        "description": "A scheme supporting eligible non-corporate, non-farm micro and small businesses through loans from participating lending institutions.",
        "eligibility": "Eligibility depends on the applicant, proposed business, lender assessment and current loan rules.",
        "age": "Applicant conditions depend on the lender and applicable scheme requirements.",
        "documents": [
            "Identity and address proof",
            "Business details or business plan, as required",
            "Bank and financial details requested by the lender",
            "Other lender-specific documents"
        ],
        "benefits": "Access to eligible business loans under applicable MUDRA loan categories. Approval is not guaranteed.",
        "application": "Review the official MUDRA website and approach an eligible participating lending institution.",
        "url": "https://www.mudra.org.in/"
    },
    {
        "name": "Pradhan Mantri Ujjwala Yojana (PMUY)",
        "aliases": "Ujjwala LPG gas cylinder cooking gas women household connection",
        "category": "Household and LPG",
        "description": "A scheme intended to provide LPG connections to eligible adult women from poor households under applicable conditions.",
        "eligibility": "Applicants must meet the household, eligibility and documentation requirements. Check the current official rules.",
        "age": "The applicant must be an adult woman and satisfy the current scheme requirements.",
        "documents": [
            "Identity and address details",
            "Ration card or other family-composition evidence, as required",
            "Bank account details",
            "Declaration and other documents specified by the scheme"
        ],
        "benefits": "LPG connection assistance under the scheme's current provisions and applicable conditions.",
        "application": "Visit the official PMUY website and follow its instructions for applying through an LPG distributor.",
        "url": "https://www.pmuy.gov.in/"
    },
    {
        "name": "National Means-cum-Merit Scholarship Scheme",
        "aliases": "NMMSS means merit scholarship school students class 8",
        "category": "School Education",
        "description": "A scholarship scheme intended to support eligible meritorious students from economically weaker sections in continuing their education.",
        "eligibility": "Students must satisfy the current examination, school, income and other scheme-specific conditions.",
        "age": "Check the latest official guidelines for applicable class and examination requirements.",
        "documents": [
            "Previous class marksheets",
            "Income certificate",
            "School and student details",
            "Bank account details",
            "Other documents specified in the current guidelines"
        ],
        "benefits": "Scholarship support according to the current scheme guidelines.",
        "application": "Check the official education department and National Scholarship Portal announcements for current application details.",
        "url": "https://scholarships.gov.in/"
    }
]

# =====================================================
# 2. TRAIN THE ML QUESTION-INTENT CLASSIFIER
# =====================================================

TRAINING_DATA = [
    ("Who can apply for this scheme?", "eligibility"),
    ("Am I eligible for PM KISAN?", "eligibility"),
    ("What are the eligibility criteria?", "eligibility"),
    ("Can a student apply for a scholarship?", "eligibility"),
    ("Who qualifies for government assistance?", "eligibility"),
    ("What conditions must I satisfy?", "eligibility"),
    ("Is this scheme available to farmers?", "eligibility"),
    ("What documents are required?", "documents"),
    ("Which certificates should I submit?", "documents"),
    ("Do I need an income certificate?", "documents"),
    ("What papers are needed for registration?", "documents"),
    ("Is Aadhaar required?", "documents"),
    ("What proof must I provide?", "documents"),
    ("What benefits does this scheme provide?", "benefits"),
    ("How much financial assistance is available?", "benefits"),
    ("What is the scholarship amount?", "benefits"),
    ("How much does PM KISAN pay annually?", "benefits"),
    ("What financial help is provided?", "benefits"),
    ("How can I apply for this scheme?", "application"),
    ("Where do I register online?", "application"),
    ("Explain the application process", "application"),
    ("Which website should I use?", "application"),
    ("How do I submit my application?", "application"),
    ("Tell me the steps to apply", "application"),
    ("What is the age limit?", "age"),
    ("What is the minimum age to apply?", "age"),
    ("Can an 18 year old apply?", "age"),
    ("What is the maximum age requirement?", "age"),
    ("How old must an applicant be?", "age"),
    ("What age group is covered?", "age"),
    ("Tell me about this government scheme", "overview"),
    ("What is PM KISAN?", "overview"),
    ("Give me an overview of the scheme", "overview"),
    ("Explain this programme", "overview")
]

X_train = [item[0] for item in TRAINING_DATA]
y_train = [item[1] for item in TRAINING_DATA]

intent_model = Pipeline([
    ("tfidf", TfidfVectorizer(
        lowercase=True,
        ngram_range=(1, 2),
        sublinear_tf=True
    )),
    ("classifier", LogisticRegression(
        max_iter=1000,
        random_state=42
    ))
])

intent_model.fit(X_train, y_train)
print("ML intent classifier trained successfully!")

# =====================================================
# 3. BUILD THE SCHEME RETRIEVAL INDEX
# =====================================================

def build_document(scheme):
    return " ".join([
        scheme["name"],
        scheme["aliases"],
        scheme["category"],
        scheme["description"],
        scheme["eligibility"],
        scheme["age"],
        " ".join(scheme["documents"]),
        scheme["benefits"],
        scheme["application"]
    ])

SCHEME_DOCUMENTS = [build_document(s) for s in SCHEMES]

scheme_vectorizer = TfidfVectorizer(
    lowercase=True,
    stop_words="english",
    ngram_range=(1, 2),
    sublinear_tf=True
)

SCHEME_MATRIX = scheme_vectorizer.fit_transform(SCHEME_DOCUMENTS)


def detect_intent(question):
    q = question.lower()

    # Explicit keywords improve common questions.
    if any(word in q for word in [
        "document", "documents", "certificate", "certificates",
        "paperwork", "papers", "aadhaar", "proof"
    ]):
        return "documents"

    if any(word in q for word in [
        "eligible", "eligibility", "qualify", "qualifies",
        "who can apply", "criteria"
    ]):
        return "eligibility"

    if any(word in q for word in [
        "age", "years old", "year old", "senior citizen", "how old"
    ]):
        return "age"

    if any(word in q for word in [
        "benefit", "benefits", "amount", "money", "financial",
        "how much", "payment", "pays"
    ]):
        return "benefits"

    if any(word in q for word in [
        "apply", "application", "register", "registration",
        "enrol", "enroll", "website"
    ]):
        return "application"

    return intent_model.predict([question])[0]


# =====================================================
# 4. FIND THE BEST-MATCHING SCHEME
# =====================================================

def find_best_scheme(question, category):
    if category == "All categories":
        candidates = list(range(len(SCHEMES)))
    else:
        candidates = [
            i for i, scheme in enumerate(SCHEMES)
            if scheme["category"] == category
        ]

    if not candidates:
        return None, 0.0

    query_vector = scheme_vectorizer.transform([question])
    scores = cosine_similarity(
        query_vector,
        SCHEME_MATRIX[candidates]
    ).flatten()

    best_position = int(scores.argmax())
    best_index = candidates[best_position]

    return SCHEMES[best_index], float(scores[best_position])


# =====================================================
# 5. GENERATE AN ANSWER FROM SCHEME DATA
# =====================================================

def answer_question(question, category):
    question = (question or "").strip()

    if not question:
        return "Please enter a question.", ""

    scheme, score = find_best_scheme(question, category)

    if scheme is None:
        return "No schemes found in this category.", ""

    # Avoid presenting unrelated results when there is no useful match.
    if score < 0.045:
        return (
            "I could not confidently match your question to a scheme "
            "in my current dataset. Try mentioning PM-KISAN, MUDRA, "
            "Ujjwala Yojana, Ayushman Bharat, or scholarship.",
            ""
        )

    intent = detect_intent(question)

    fields = {
        "eligibility": (
            "Eligibility",
            scheme["eligibility"]
        ),
        "documents": (
            "Required documents",
            "\n".join(
                f"- {doc}" for doc in scheme["documents"]
            )
        ),
        "age": (
            "Age requirements",
            scheme["age"]
        ),
        "benefits": (
            "Benefits",
            scheme["benefits"]
        ),
        "application": (
            "How to apply",
            scheme["application"]
        ),
        "overview": (
            "Overview",
            scheme["description"]
        )
    }

    label, content = fields.get(
        intent,
        ("Overview", scheme["description"])
    )

    answer = (
        f"## {scheme['name']}\n\n"
        f"**Category:** {scheme['category']}\n\n"
        f"**{label}:**\n\n{content}\n\n"
        f"**Retrieval match score:** {score:.3f}\n\n"
        "**Important:** This prototype uses a small built-in dataset "
        "and does not verify live scheme rules. Confirm current "
        "eligibility, documents, deadlines and benefits on the "
        "official website."
    )

    sources = (
        "## Official source\n\n"
        f"[{scheme['name']}]({scheme['url']})"
    )

    return answer, sources


# =====================================================
# 6. GRADIO WEB INTERFACE
# =====================================================

categories = ["All categories"] + sorted({
    scheme["category"] for scheme in SCHEMES
})

examples = [
    ["What documents are needed for PM-KISAN?", "All categories"],
    ["Who is eligible for Ujjwala Yojana?", "All categories"],
    ["What is the age limit for Ayushman Bharat senior citizens?", "All categories"],
    ["How much benefit does PM-KISAN provide?", "All categories"],
    ["How do I apply for a MUDRA loan?", "All categories"],
    ["What documents are required for a school scholarship?", "All categories"]
]

with gr.Blocks(title="AI Government Schemes Assistant") as demo:
    gr.Markdown(
        "# 🇮🇳 AI Government Schemes Information Assistant\n\n"
        "Ask about eligibility, age limits, documents, benefits "
        "or application steps for selected Indian government schemes."
    )

    with gr.Row():
        question = gr.Textbox(
            label="Ask your question",
            placeholder="Example: What documents are needed for PM-KISAN?",
            lines=2,
            scale=4
        )

        category = gr.Dropdown(
            choices=categories,
            value="All categories",
            label="Filter by category",
            scale=2
        )

    search_button = gr.Button(
        "Find Government Scheme",
        variant="primary"
    )

    answer_output = gr.Markdown()
    source_output = gr.Markdown()

    search_button.click(
        fn=answer_question,
        inputs=[question, category],
        outputs=[answer_output, source_output]
    )

    question.submit(
        fn=answer_question,
        inputs=[question, category],
        outputs=[answer_output, source_output]
    )

    gr.Examples(
        examples=examples,
        inputs=[question, category],
        label="Try these example questions"
    )

    gr.Markdown(
        "---\n"
        "**Project note:** This is a basic retrieval-based RAG-style "
        "prototype with an ML intent classifier. It is not a trained "
        "large language model, and the built-in data may be incomplete "
        "or out of date."
    )

# =====================================================
# 7. LAUNCH THE APPLICATION
# =====================================================

demo.launch(share=True, debug=False)
```
